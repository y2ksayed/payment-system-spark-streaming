# Fraud analysis — scored payments

Reads the Parquet output of `streaming_job.py` and explores fraud scores:
score distribution, flagged rate by merchant, and amount vs probability.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("../data/output/scored_payments")
print(df.shape)
df[["transaction_id", "merchant", "amount", "fraud_probability", "is_fraud"]].head()
```

```python
# Score distribution: most payments should sit near 0
df["fraud_probability"].hist(bins=50)
plt.xlabel("fraud probability")
plt.ylabel("count")
plt.title("Fraud score distribution")
plt.show()
```

```python
# Flagged rate and average amount by merchant category
summary = (
    df.groupby("merchant")
    .agg(flagged_rate=("is_fraud", "mean"), avg_amount=("amount", "mean"), n=("amount", "size"))
    .sort_values("flagged_rate", ascending=False)
)
summary
```

```python
# Sanity check: flagged payments should skew toward higher amounts
print(df.groupby("is_fraud")["amount"].describe())
```
